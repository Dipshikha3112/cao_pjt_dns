Cell 2 — Import Stage 3.0

```python
# ============================================================
# CELL 2 — LOAD STAGE 3.0
# ============================================================

%run "step0.ipynb"

print("\nStage 3.0 environment loaded successfully.")
```

Cell 3 — Check the required data

```python
# ============================================================
# CELL 3 — VERIFY SPLIT DATA IS AVAILABLE
# ============================================================

required_variables = [
    "CISLR_ROOT",
    "split_membership_df",
    "video_mapping_df"
]

missing_variables = [
    var for var in required_variables
    if var not in globals()
]

if missing_variables:
    raise RuntimeError(
        f"Missing variables from step0.ipynb: {missing_variables}"
    )

print("Required Stage 3.0 variables are available.")
```

Cell 4 — Inspect split membership

```python
# ============================================================
# CELL 4 — INSPECT OFFICIAL SPLIT MEMBERSHIP
# ============================================================

print("Split membership columns:")
print(split_membership_df.columns.tolist())

print("\nFirst 5 rows:")
display(split_membership_df.head())
```

Cell 5 — Verify available split labels

```python
# ============================================================
# CELL 5 — VERIFY SPLIT LABELS
# ============================================================

print("=" * 80)
print("AVAILABLE SPLITS")
print("=" * 80)

print(
    split_membership_df["split"]
    .value_counts()
)
```

Cell 6 — Create train and test UID sets

```python
# ============================================================
# CELL 6 — SELECT OFFICIAL TRAIN / TEST UIDs
# ============================================================

train_uids = set(
    split_membership_df.loc[
        split_membership_df["split"] == "train",
        "uid"
    ]
)

test_uids = set(
    split_membership_df.loc[
        split_membership_df["split"] == "test",
        "uid"
    ]
)

print("Official train UIDs:", f"{len(train_uids):,}")
print("Official test UIDs :", f"{len(test_uids):,}")
```

Cell 7 — Verify no overlap

```python
# ============================================================
# CELL 7 — TRAIN / TEST OVERLAP CHECK
# ============================================================

uid_overlap = train_uids.intersection(test_uids)

print("Train/Test UID overlap:", len(uid_overlap))

if len(uid_overlap) > 0:
    print("\nERROR — TRAIN/TEST OVERLAP DETECTED")
    
    print("\nOverlapping UIDs:")
    display(
        pd.DataFrame(
            {"uid": sorted(uid_overlap)}
        ).head(20)
    )
    
    raise RuntimeError(
        "Official train/test split contains UID overlap."
    )

print("\nPASS — No train/test UID overlap.")
```

Cell 8 — Verify every UID belongs to exactly one split

```python
# ============================================================
# CELL 8 — VERIFY UNIQUE SPLIT MEMBERSHIP
# ============================================================

split_counts = (
    split_membership_df
    .groupby("uid")["split"]
    .nunique()
)

multiple_split_uids = split_counts[
    split_counts > 1
]

print(
    "UIDs assigned to multiple splits:",
    len(multiple_split_uids)
)

if len(multiple_split_uids) > 0:
    print("\nERROR — UID assigned to multiple splits.")
    
    display(
        multiple_split_uids
        .head(20)
    )
    
    raise RuntimeError(
        "Some UIDs belong to multiple splits."
    )

print("\nPASS — Every UID belongs to only one split.")
```

Cell 9 — Connect split membership to video mapping

```python
# ============================================================
# CELL 9 — BUILD VERIFIED VIDEO SPLIT TABLE
# ============================================================

split_video_df = video_mapping_df.merge(
    split_membership_df[
        ["uid", "split"]
    ],
    on="uid",
    how="left",
    validate="one_to_one"
)

print("Verified video split table created.")
print("Shape:", split_video_df.shape)

display(
    split_video_df.head()
)
```

Cell 10 — Check for missing split assignments

```python
# ============================================================
# CELL 10 — CHECK MISSING SPLIT ASSIGNMENTS
# ============================================================

missing_split = split_video_df[
    split_video_df["split"].isna()
]

print(
    "Videos without split assignment:",
    len(missing_split)
)

if len(missing_split) > 0:
    print("\nERROR — Some videos have no split assignment.")
    display(missing_split.head(20))
    
    raise RuntimeError(
        "Missing train/test split assignments detected."
    )

print("\nPASS — Every mapped video has a split assignment.")
```

Cell 11 — Create final train and test tables

```python
# ============================================================
# CELL 11 — CREATE OFFICIAL TRAIN / TEST TABLES
# ============================================================

train_df = split_video_df[
    split_video_df["split"] == "train"
].copy()

test_df = split_video_df[
    split_video_df["split"] == "test"
].copy()

print("=" * 80)
print("OFFICIAL DATA SPLIT")
print("=" * 80)

print("Train videos:", f"{len(train_df):,}")
print("Test videos :", f"{len(test_df):,}")
print(
    "Total videos:",
    f"{len(train_df) + len(test_df):,}"
)
```

Cell 12 — Verify counts

```python
# ============================================================
# CELL 12 — FINAL SPLIT COUNT VERIFICATION
# ============================================================

total_mapped = len(split_video_df)
total_split = len(train_df) + len(test_df)

print("Total mapped videos :", f"{total_mapped:,}")
print("Train videos        :", f"{len(train_df):,}")
print("Test videos         :", f"{len(test_df):,}")
print("Train + Test        :", f"{total_split:,}")

if total_mapped != total_split:
    raise RuntimeError(
        "Train + Test count does not equal total mapped videos."
    )

print("\nPASS — All videos are accounted for.")
```

Cell 13 — Verify class labels

```python
# ============================================================
# CELL 13 — VERIFY GLOSS LABELS
# ============================================================

print("Missing gloss labels:")
print(
    split_video_df["gloss"].isna().sum()
)

print("\nUnique glosses in complete dataset:")
print(
    split_video_df["gloss"].nunique()
)

print("\nUnique glosses in train:")
print(
    train_df["gloss"].nunique()
)

print("\nUnique glosses in test:")
print(
    test_df["gloss"].nunique()
)
```

Cell 14 — Check test videos are actually official test videos

```python
# ============================================================
# CELL 14 — VERIFY OFFICIAL TEST SET
# ============================================================

official_test_file = "/home/dipshikha/Music/cao/test.csv"

official_test_df = pd.read_csv(
    official_test_file
)

official_test_uids = set(
    official_test_df["uid"]
    .astype(str)
    .str.strip()
)

selected_test_uids = set(
    test_df["uid"]
    .astype(str)
    .str.strip()
)

missing_from_selected = (
    official_test_uids - selected_test_uids
)

unexpected_in_selected = (
    selected_test_uids - official_test_uids
)

print("Official test.csv UIDs :", f"{len(official_test_uids):,}")
print("Selected test UIDs     :", f"{len(selected_test_uids):,}")
print(
    "Missing official UIDs  :",
    f"{len(missing_from_selected):,}"
)
print(
    "Unexpected test UIDs   :",
    f"{len(unexpected_in_selected):,}"
)
```

Cell 15 — Final official test verification

```python
# ============================================================
# CELL 15 — OFFICIAL TEST VERIFICATION RESULT
# ============================================================

if missing_from_selected:
    raise RuntimeError(
        "Some official test.csv UIDs are missing from the "
        "Stage 3 test set."
    )

if unexpected_in_selected:
    raise RuntimeError(
        "Stage 3 test set contains UIDs that are not in "
        "official test.csv."
    )

print("=" * 80)
print("PASS — STAGE 3 TEST SET MATCHES OFFICIAL test.csv")
print("=" * 80)
```
