```python
# ============================================================
# CISLR — STEP 5
# SPLIT VERIFICATION & DATA LEAKAGE ANALYSIS
# ============================================================
#
# Purpose:
#   Verify that the official CISLR test set is correctly
#   separated from the training data and identify possible
#   sources of data leakage.
#
# Checks:
#   1. Official test-set membership
#   2. Train/test UID overlap
#   3. Duplicate UID entries
#   4. Duplicate video content
#   5. UID ↔ filename consistency
#   6. Potential data leakage
#
# Inputs:
#   Step 2 -> video_mapping.csv
#   CISLR  -> test.csv
#
# Outputs:
#   split_verification.csv
#   leakage_report.csv
# ============================================================

print("=" * 80)
print("CISLR — STEP 5: SPLIT VERIFICATION")
print("=" * 80)
```

CELL 3 — Load Step 2 mapping

```python
# ============================================================
# CELL 3 — LOAD VIDEO MAPPING
# ============================================================
import pandas as pd
from pathlib import Path
cislr_research="/home/dipshikha/Music/cao/CISLR_RESEARCH"
mapping_file = (
    Path(cislr_research)
    / "processed"
    / "video_mapping.csv"
)

video_mapping_df = pd.read_csv(mapping_file)

print("=" * 80)
print("VIDEO MAPPING LOADED")
print("=" * 80)

print("File:", mapping_file)
print("Rows:", len(video_mapping_df))

display(
    video_mapping_df.head()
)
```

CELL 4 — Load official test.csv

```python
# ============================================================
# CELL 4 — LOAD OFFICIAL TEST SET
# ============================================================

test_file = "/home/dipshikha/Music/cao/test.csv"

official_test_df = pd.read_csv(test_file)

print("=" * 80)
print("OFFICIAL CISLR TEST SET")
print("=" * 80)

print("File:", test_file)
print("Rows:", len(official_test_df))

print("\nColumns:")
print(official_test_df.columns.tolist())

display(
    official_test_df.head()
)
```

CELL 5 — Verify required columns

```python
# ============================================================
# CELL 5 — VERIFY REQUIRED COLUMNS
# ============================================================

required_mapping_columns = [
    "uid",
    "video_path"
]

required_test_columns = [
    "uid"
]

missing_mapping_columns = [
    col
    for col in required_mapping_columns
    if col not in video_mapping_df.columns
]

missing_test_columns = [
    col
    for col in required_test_columns
    if col not in official_test_df.columns
]

if missing_mapping_columns:
    raise ValueError(
        f"Missing columns in video_mapping.csv: "
        f"{missing_mapping_columns}"
    )

if missing_test_columns:
    raise ValueError(
        f"Missing columns in test.csv: "
        f"{missing_test_columns}"
    )

print("PASS — Required columns are available.")
```

CELL 6 — Clean UID values

```python
# ============================================================
# CELL 6 — NORMALIZE UID VALUES FOR COMPARISON
# ============================================================

video_mapping_df["uid_clean"] = (
    video_mapping_df["uid"]
    .astype(str)
    .str.strip()
)

official_test_df["uid_clean"] = (
    official_test_df["uid"]
    .astype(str)
    .str.strip()
)

print("UID normalization complete.")

print(
    "Unique UIDs in complete dataset:",
    video_mapping_df["uid_clean"].nunique()
)

print(
    "Unique UIDs in official test:",
    official_test_df["uid_clean"].nunique()
)
```

CELL 7 — Verify official test UIDs exist

```python
# ============================================================
# CELL 7 — OFFICIAL TEST UID EXISTENCE
# ============================================================

all_dataset_uids = set(
    video_mapping_df["uid_clean"]
)

official_test_uids = set(
    official_test_df["uid_clean"]
)

test_uids_missing_from_dataset = sorted(
    official_test_uids - all_dataset_uids
)

print("=" * 80)
print("OFFICIAL TEST UID EXISTENCE CHECK")
print("=" * 80)

print(
    "Official test UIDs:",
    len(official_test_uids)
)

print(
    "Missing from dataset:",
    len(test_uids_missing_from_dataset)
)

if len(test_uids_missing_from_dataset) == 0:

    print("\nPASS — Every official test UID exists in the dataset.")

else:

    print("\nFAIL — Test UIDs missing from dataset:")

    for uid in test_uids_missing_from_dataset[:50]:
        print(uid)
```

CELL 8 — Construct the official train set

```python
# ============================================================
# CELL 8 — CONSTRUCT OFFICIAL TRAIN SET
# ============================================================

train_df = video_mapping_df[
    ~video_mapping_df["uid_clean"].isin(
        official_test_uids
    )
].copy()

test_df = video_mapping_df[
    video_mapping_df["uid_clean"].isin(
        official_test_uids
    )
].copy()

print("=" * 80)
print("OFFICIAL TRAIN / TEST PARTITION")
print("=" * 80)

print("Complete dataset :", len(video_mapping_df))
print("Train samples     :", len(train_df))
print("Test samples      :", len(test_df))

print(
    "Train + Test      :",
    len(train_df) + len(test_df)
)
```

CELL 9 — Train/test UID overlap

```python
# ============================================================
# CELL 9 — TRAIN / TEST UID OVERLAP
# ============================================================

train_uids = set(
    train_df["uid_clean"]
)

test_uids = set(
    test_df["uid_clean"]
)

train_test_uid_overlap = sorted(
    train_uids & test_uids
)

print("=" * 80)
print("TRAIN / TEST UID OVERLAP")
print("=" * 80)

print(
    "Train unique UIDs:",
    len(train_uids)
)

print(
    "Test unique UIDs:",
    len(test_uids)
)

print(
    "Overlapping UIDs:",
    len(train_test_uid_overlap)
)

if len(train_test_uid_overlap) == 0:

    print("\nPASS — No UID appears in both train and test.")

else:

    print("\nFAIL — UID overlap detected:")

    for uid in train_test_uid_overlap[:50]:
        print(uid)
```

CELL 10 — Duplicate UIDs in complete dataset

```python
# ============================================================
# CELL 10 — DUPLICATE UID CHECK
# ============================================================

uid_counts = (
    video_mapping_df["uid_clean"]
    .value_counts()
)

duplicate_uids = uid_counts[
    uid_counts > 1
]

print("=" * 80)
print("DUPLICATE UID CHECK")
print("=" * 80)

print(
    "Duplicate UID values:",
    len(duplicate_uids)
)

if len(duplicate_uids) == 0:

    print("\nPASS — No duplicate UIDs.")

else:

    print("\nWARNING — Duplicate UIDs detected:")

    display(
        duplicate_uids.to_frame(
            "occurrences"
        )
    )
```

CELL 11 — Check physical video path overlap

```python
# ============================================================
# CELL 11 — PHYSICAL VIDEO PATH OVERLAP
# ============================================================

train_paths = set(
    train_df["video_path"]
    .astype(str)
)

test_paths = set(
    test_df["video_path"]
    .astype(str)
)

path_overlap = sorted(
    train_paths & test_paths
)

print("=" * 80)
print("PHYSICAL VIDEO PATH OVERLAP")
print("=" * 80)

print(
    "Overlapping video paths:",
    len(path_overlap)
)

if len(path_overlap) == 0:

    print(
        "\nPASS — No physical video file appears in both splits."
    )

else:

    print(
        "\nFAIL — Same physical video appears in both splits:"
    )

    for path in path_overlap[:50]:
        print(path)
```

CELL 12 — Check duplicate video content across train/test

```python
# ============================================================
# CELL 12 — EXACT VIDEO CONTENT LEAKAGE CHECK
# ============================================================

import hashlib
from pathlib import Path

def calculate_sha256(file_path, chunk_size=1024 * 1024):

    sha256 = hashlib.sha256()

    with open(file_path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            sha256.update(chunk)

    return sha256.hexdigest()
```

CELL 13 — Hash train and test videos

```python
# ============================================================
# CELL 13 — CALCULATE TRAIN / TEST VIDEO HASHES
# ============================================================

train_hashes = {}
test_hashes = {}

print("Hashing TRAIN videos...")

for i, path in enumerate(train_paths):

    path = Path(path)

    if path.exists():

        file_hash = calculate_sha256(path)

        train_hashes.setdefault(
            file_hash,
            []
        ).append(path)

    if (i + 1) % 500 == 0:

        print(
            f"Train: {i + 1}/{len(train_paths)}"
        )


print("\nHashing TEST videos...")

for i, path in enumerate(test_paths):

    path = Path(path)

    if path.exists():

        file_hash = calculate_sha256(path)

        test_hashes.setdefault(
            file_hash,
            []
        ).append(path)

    if (i + 1) % 500 == 0:

        print(
            f"Test: {i + 1}/{len(test_paths)}"
        )

print("\nHashing complete.")
```

CELL 14 — Detect train/test content duplicates

```python
# ============================================================
# CELL 14 — TRAIN / TEST CONTENT OVERLAP
# ============================================================

train_hash_set = set(
    train_hashes.keys()
)

test_hash_set = set(
    test_hashes.keys()
)

common_hashes = (
    train_hash_set &
    test_hash_set
)

print("=" * 80)
print("TRAIN / TEST CONTENT OVERLAP")
print("=" * 80)

print(
    "Unique train video hashes:",
    len(train_hash_set)
)

print(
    "Unique test video hashes:",
    len(test_hash_set)
)

print(
    "Identical content groups across splits:",
    len(common_hashes)
)
```

CELL 15 — Display potential content leakage

```python
# ============================================================
# CELL 15 — CONTENT LEAKAGE REPORT
# ============================================================

if len(common_hashes) == 0:

    print(
        "PASS — No byte-identical videos were found "
        "across train and test."
    )

else:

    print(
        "WARNING — Potential content leakage detected."
    )

    print(
        "\nIdentical train/test video groups:"
    )

    for file_hash in common_hashes:

        print("\nSHA-256:", file_hash)

        print("TRAIN:")

        for path in train_hashes[file_hash]:
            print("  ", path)

        print("TEST:")

        for path in test_hashes[file_hash]:
            print("  ", path)
```

CELL 16 — Check official test label consistency

```python
# ============================================================
# CELL 16 — TEST LABEL CONSISTENCY
# ============================================================

dataset_labels = (
    video_mapping_df[
        ["uid_clean", "gloss"]
    ]
    .drop_duplicates("uid_clean")
)

test_labels = (
    official_test_df[
        ["uid_clean", "gloss"]
    ]
    .drop_duplicates("uid_clean")
)

label_check_df = test_labels.merge(
    dataset_labels,
    on="uid_clean",
    how="left",
    suffixes=(
        "_test_csv",
        "_dataset"
    )
)

label_mismatch_df = label_check_df[
    label_check_df["gloss_test_csv"]
    !=
    label_check_df["gloss_dataset"]
].copy()

print("=" * 80)
print("OFFICIAL TEST LABEL CONSISTENCY")
print("=" * 80)

print(
    "Label mismatches:",
    len(label_mismatch_df)
)

if len(label_mismatch_df) == 0:

    print(
        "\nPASS — Official test labels match dataset labels."
    )

else:

    print(
        "\nFAIL — Label mismatch detected:"
    )

    display(label_mismatch_df)
```

CELL 17 — Potential leakage summary

```python
# ============================================================
# CELL 17 — POTENTIAL LEAKAGE SUMMARY
# ============================================================

potential_leakage = {
    "train_test_uid_overlap": len(
        train_test_uid_overlap
    ),

    "train_test_path_overlap": len(
        path_overlap
    ),

    "train_test_identical_content": len(
        common_hashes
    ),

    "duplicate_uids_in_dataset": len(
        duplicate_uids
    ),

    "official_test_uids_missing": len(
        test_uids_missing_from_dataset
    ),

    "test_label_mismatches": len(
        label_mismatch_df
    )
}

leakage_summary_df = pd.DataFrame(
    potential_leakage.items(),
    columns=[
        "check",
        "count"
    ]
)

print("=" * 80)
print("POTENTIAL LEAKAGE SUMMARY")
print("=" * 80)

display(leakage_summary_df)
```

CELL 18 — Final PASS / FAIL

```python
# ============================================================
# CELL 18 — STEP 5 FINAL STATUS
# ============================================================

critical_failures = (
    len(test_uids_missing_from_dataset)
    +
    len(train_test_uid_overlap)
    +
    len(path_overlap)
    +
    len(common_hashes)
    +
    len(label_mismatch_df)
)

print("=" * 80)
print("STEP 5 — FINAL STATUS")
print("=" * 80)

if critical_failures == 0:

    print("STEP 5 STATUS: PASS")
    print()
    print("✓ Official test UIDs are valid")
    print("✓ No train/test UID overlap")
    print("✓ No physical video path overlap")
    print("✓ No byte-identical video content across splits")
    print("✓ Official test labels are consistent")

else:

    print("STEP 5 STATUS: REVIEW REQUIRED")
    print()
    print(
        "Critical issues detected:",
        critical_failures
    )

print("=" * 80)
```
